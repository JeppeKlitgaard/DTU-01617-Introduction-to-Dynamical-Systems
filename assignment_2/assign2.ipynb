### Problem 1c

```python
# Phase portrait for Duffing system
import numpy as np
import matplotlib.pyplot as plt

def f(x, mu=1.0):
    x1, x2 = x
    return np.array([x2, mu*x1 - x1**3])

def plot_field(mu=1.0, x1lim=(-3,3), x2lim=(-3,3), density=30):
    x1 = np.linspace(x1lim[0], x1lim[1], density)
    x2 = np.linspace(x2lim[0], x2lim[1], density)
    X1, X2 = np.meshgrid(x1, x2)
    U = X2
    V = mu*X1 - X1**3
    N = np.sqrt(U**2 + V**2)
    U2, V2 = U/N, V/N  # normalize for nicer quiver plotting

    plt.figure(figsize=(6,6), dpi=250)
    plt.quiver(X1, X2, U2, V2, angles='xy')
    plt.xlabel('$x_1$'); plt.ylabel('$x_2$')
    plt.title(f'Duffing vector field, mu={mu}')
    plt.axhline(0,color='k',linewidth=0.5); 
    plt.axvline(0,color='k',linewidth=0.5)
    plt.xlim(x1lim); plt.ylim(x2lim)
    plt.gca().set_aspect('equal', 'box')
    #plt.show()
```

```python
# Example: mu=1 => mu>0 => (three equilibria)
mu = 1.0
plot_field(mu=mu, x1lim=(-3,3), x2lim=(-3,3))

plt.annotate('saddle', xy=(0, 0), xytext=(0.5, 1), arrowprops=dict(facecolor='black', arrowstyle='->'), backgroundcolor='lightgray', fontsize=10)
plt.scatter([0], [0], color='red')  # saddle point at origin
plt.annotate('linear center', xy=(mu**0.5, 0), xytext=(1.5, -1), arrowprops=dict(facecolor='black', arrowstyle='->'), backgroundcolor='lightgray', fontsize=10)
plt.scatter([mu**0.5], [0], color='blue')  # center at (sqrt(mu), 0)
plt.annotate('linear center', xy=(-mu**0.5, 0), xytext=(-2, -1), arrowprops=dict(facecolor='black', arrowstyle='->'), backgroundcolor='lightgray', fontsize=10)
plt.scatter([-mu**0.5], [0], color='blue')  # center at (-sqrt(mu), 0)
```

```python
# Example: mu=-1 => mu<0 => (one center)
plot_field(mu=-1.0, x1lim=(-2,2), x2lim=(-2,2))

plt.annotate('linear center (orbit)', xy=(0, 0), xytext=(0.5, 1), arrowprops=dict(facecolor='black', arrowstyle='->'), backgroundcolor='lightgray', fontsize=10)
plt.scatter([0], [0], color='red')  # saddle point at origin
```

### Problem 1e

```python
import numpy as np
import matplotlib.pyplot as plt

mu = 1.0
x1_vals = np.linspace(0, np.sqrt(2*mu), 200)
h_vals = x1_vals * np.sqrt(np.maximum(0, mu - 0.5*x1_vals**2))

# vector field (small region)
x1 = np.linspace(-3, 3, 40); x2 = np.linspace(-2, 2, 30)
X1, X2 = np.meshgrid(x1, x2)
U = X2
V = mu*X1 - X1**3
N = np.sqrt(U**2 + V**2)

# plotting
plt.figure(figsize=(10,4), dpi=250)

plt.quiver(X1, X2, U/N, V/N, angles='xy')
plt.plot(x1_vals, h_vals, 'green', linewidth=2, label='$h(x_1)^+$ orbit (first quadrant)', alpha=0.7)
plt.plot(x1_vals, -h_vals, 'lightgreen', linewidth=2, label='$h(x_1)^-$ orbit (fourth quadrant)', alpha=0.7)

#plt.scatter([0, np.sqrt(mu), -np.sqrt(mu)],[0,0,0], color='k')
plt.scatter([0], [0], color='red')  # saddle point at origin
plt.scatter([mu**0.5], [0], color='blue')  # center at (sqrt(mu), 0)
plt.scatter([-mu**0.5], [0], color='blue')  # center at (-sqrt(mu), 0)

plt.xlim(-3,3); plt.ylim(-2,2)
plt.gca().set_aspect('equal', 'box')
plt.legend(); plt.title('Duffing vector field - analytic homoclinic orbit for mu=1')
plt.show()
```
